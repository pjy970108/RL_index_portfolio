```python
# 1. 인덱스를 datetime 형식으로 변환
grpo_best.index = pd.to_datetime(grpo_best.index)
```

```python
grpo_best['year'] = grpo_best.index.year
```

```python

# 3. 연도별 수익률 합계
yearly_return = grpo_best.groupby('year')['Model'].sum()
```

```python
# 4. 수익이 난 해 / 손실이 난 해 개수
profitable_years = (yearly_return > 0).sum()
unprofitable_years = (yearly_return <= 0).sum()

print(f"Profitable years: {profitable_years}, Unprofitable years: {unprofitable_years}")
```
