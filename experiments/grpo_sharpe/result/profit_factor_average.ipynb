```python
for col in test_return.columns:
    model = test_return[col]
    gross_profit = model[model >= 0].sum()
    gross_loss = model[model < 0].sum()
    profit_factor = gross_profit / abs(gross_loss) if gross_loss != 0 else np.nan
    print(f"gross_profit for {col}: {gross_profit:.4f}")
    print(f"gross_loss for {col}: {gross_loss:.4f}")
    print(f"Profit Factor for {col}: {profit_factor:.4f}")
    print("profitabel_months:", (test_return[col] > 0).sum())
    print("unprofitabel_months:", (test_return[col] < 0).sum())
    print("-"*50)
```

```python
for col in test_return.columns:
    model = test_return[col]
    model.index = pd.to_datetime(model.index)
    
    # Series → DataFrame 변환
    model_df = model.to_frame(name='return')
    model_df['year'] = model_df.index.year

    yearly_return = model_df.groupby('year')['return'].sum()
    
    profitable_years = (yearly_return >= 0).sum()
    unprofitable_years = (yearly_return < 0).sum()
    
    print(f"Profitable years for {col}: {profitable_years}, Unprofitable years: {unprofitable_years}")
```
