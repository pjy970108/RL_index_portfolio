```python
import pandas as pd
import numpy as np
import robustsharpe as rs
```

```python
# test_return = pd.read_csv("./benchmark/0.003_benchmark.csv", index_col=0)
test_return = pd.read_csv("./benchmark_del/0.003_benchmark.csv", index_col=0)
```

```python
test_return.columns
```

```python
benchmark_returns = np.array(test_return["equal_strategy"])

col_list = ['risk_parity', 'min_var', 'max_sharpe', 'paa', 'equal_asset_weight']
for col in col_list:
    my_strategy_returns = np.array(test_return[col])
    
    returns = np.stack([my_strategy_returns, benchmark_returns], axis=1)  # shape: (T, 2)
    SRs, diff, ci, pval, se, d = rs.bootstrap_inference(
        returns=returns,
        block_size=4,     # 논문 추천값 (T=120 기준)
        alpha=0.05,
        M=1000
    )
    print(f"Results for {col}:")
    print("Sharpe ratio (벤치마크, 내):", SRs)
    print("Sharpe ratio 차이:", diff)
    print("95% 신뢰구간:", ci)
    print("t-통계량", d)
    print("p-value:", pval)
    print("-"*50)
```
