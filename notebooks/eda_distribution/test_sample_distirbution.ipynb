```python
import matplotlib.pyplot as plt
import numpy as np
# 전체 샘플 히스토그램
plt.figure(figsize=(12, 8))
plt.hist(loaded_samples, bins=20, alpha=0.6, label=f"rewards")

plt.title("Histogram of Normalized Rewards (Dirichlet Samples)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# 전체 샘플 히스토그램
plt.figure(figsize=(12, 8))
plt.hist(loaded_rewards, bins=20, alpha=0.6, label=f"rewards")

plt.title("Histogram of Normalized Rewards (Dirichlet Samples)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)
plt.show()
```

```python


# 디리클레 분포 샘플 (가정: 이전에 저장된 데이터)

# 각 asset별 histogram 시각화
fig, axs = plt.subplots(3, 2, figsize=(12, 10))
axs = axs.flatten()

for i in range(loaded_samples.shape[1]):
    axs[i].hist(loaded_samples[:, i], bins=50, alpha=0.7)
    axs[i].set_title(f"Asset {i+1}")

plt.tight_layout()
plt.show()
```
